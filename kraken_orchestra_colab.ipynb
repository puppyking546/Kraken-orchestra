### 3. Upload your song

```python
from google.colab import files
up = files.upload()
song = list(up)[0]
print("Got:", song)
```

### 4. Settings and run
Test a short clip first. Set `DURATION = None` to do the whole song.

```python
START = 30        # start second
DURATION = 45     # seconds to process, or None for the whole song
OTHER_MOTORS = 5  # motors for guitar/piano/synth harmony
ROOM = True       # add a little reverb

out = "kraken_out.wav"
cmd = f'python kraken_orchestra_audio.py "../{song}" -o {out} --start {START} --other-motors {OTHER_MOTORS}'
if DURATION: cmd += f" --duration {DURATION}"
if ROOM: cmd += " --room"
!{cmd}
```

### 5. Listen and download

```python
from IPython.display import Audio, display
display(Audio(out))
files.download(out)
```
